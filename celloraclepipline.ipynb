```python
import scanpy as sc
import pandas as pd
import numpy as np

# Load the Stage 18 
practicedata = sc.read_h5ad("Stage18_Data.h5ad")
```

```python
adata_pax6_wt = practicedata[practicedata.obs['orig.ident'] == 'pax6_WT'].copy()
adata_pax6_mut = practicedata[practicedata.obs['orig.ident'] == 'pax6_Mutant'].copy()
```

```python
adata_pax6_wt_6_9 = adata_pax6_wt[
    adata_pax6_wt.obs['seurat_clusters'].isin([6, 9])
].copy()
adata_pax6_mut_6_9 = adata_pax6_mut[
    adata_pax6_mut.obs['seurat_clusters'].isin([6, 9])
].copy()
```

```python
base_GRN = co.data.load_xenopus_tropicalis_promoter_base_GRN(
    version="xenTro3_CisBPv2_fpr2"
)
base_GRN.head()
```

```python
print("Metadata columns :", list(adata_pax6_wt_6_9.obs.columns))
print("Dimensional reduction: ", list(adata_pax6_wt_6_9.obsm.keys()))
```

```python
adata = adata_pax6_wt_6_9.copy()
adata.X = adata.raw[:, adata.var_names].X.copy()
oracle = co.Oracle()
```

```python
print("Metadata columns :", list(adata.obs.columns))
print("Dimensional reduction: ", list(adata.obsm.keys()))
```

```python
oracle.import_anndata_as_normalized_count(
    adata=adata,

    # Seurat clusters 6 and 9
    cluster_column_name="seurat_clusters",

    # UMAP coordinates from your dataset
    embedding_name="X_umap"
)
```

```python
oracle.import_TF_data(TF_info_matrix=base_GRN)
```

```python
oracle.perform_PCA()
plt.plot(np.cumsum(oracle.pca.explained_variance_ratio_)[:100])
n_comps = np.where(np.diff(np.diff(np.cumsum(oracle.pca.explained_variance_ratio_))>0.002))[0][0]
plt.axvline(n_comps, c="k")
plt.show()
print(n_comps)
n_comps = min(n_comps, 50)
```
